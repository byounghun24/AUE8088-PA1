```python
import matplotlib.pyplot as plt
```

```python
# 모델 크기 (단위: M)과 검증 정확도 (단위: %)
model_size = [4.3,  6.8,   8.0,   11.0,  17.9,  28.8,  41.2,  64.3]
val_acc    = [38.4, 40.12, 40.36, 41.12, 41.44, 42.72, 43.44, 44.12]

# 일부 모델 이름 표시 (필요한 경우만)
model_labels = ['b0', 'b1', 'b2', 'b3', 'b4', 'b5', 'b6', 'b7']

# 그래프 그리기
plt.figure(figsize=(8, 6))
plt.plot(model_size, val_acc, marker='o', label='EfficientNet')

# 각 점에 라벨 추가
for i, label in enumerate(model_labels):
    if label:
        plt.text(model_size[i], val_acc[i] + 0.2, label, ha='center')

# 라벨 및 제목 설정
plt.title('[ Size-Accuracy Trade-off ]', fontsize=14)
plt.xlabel('Model Size (M)', fontsize=12)
plt.ylabel('Val. Acc (%)', fontsize=12)

plt.xlim(0, 70)
plt.ylim(38.0, 45.0)

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
